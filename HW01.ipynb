## Question 6: Median value of horsepower
1. Find the median value of the horsepower column in the dataset.
2. Next, calculate the most frequent value of the same horsepower column.
3. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
4. Now, calculate the median value of horsepower once again.

```python
# First the horsepower median 
median = df["horsepower"].median()

#Second the mode 
mode = df["horsepower"].mode()

#Third, fill te null valuses in horsepower
df["horsepower"] = df["horsepower"].fillna(mode)

# Forth, re calculate 
median_v2 = df["horsepower"].median()

# print to see 
print("This is the median befor the fillna method " , median)
print("This is the median after the fillna method " , median_v2)


```

## Question 7 : Sum of weights
1. Select all the cars from Asia
2. Select only columns vehicle_weight and model_year
3. Select the first 7 values
4. Get the underlying NumPy array. Let's call it X.
5. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
Invert XTX.
5. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
6. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
7. What's the sum of all the elements of the result?

```python
# Step 1 , 2 and 3 
asia = df[df["origin"] == "Asia"][["vehicle_weight","model_year"]].head(7)

# Getting X and X transpose , step 4,5
X= asia.to_numpy()
XTX = X.T @ X
invert_XTX = np.linalg.inv(XTX)

# Creating y , step 6
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Steps 7, 8
w = (invert_XTX @ X.T) @ y 
print("The sum of w is ", w.sum())
```
